### Pregunta 6 **<font color="blue" size="+2">EG</font>**

Python dispone de un idiom muy útil conocido como list comprehensions. ¿Puedes explicar en qué consiste? Indicad
**dos ejemplos** de código que utilicen este idiom y documentad cada parte de código relevante.

**Respuesta:**

Este idiom nos permite crear listas de manera concisa y eficiente utilizando solo una línea de código, aplicando una expresión a una secuencia (como una lista o un rango) y opcionalmente filtrando los elementos con una condición.
A nivel de ejemplos tenemos:


````python
# Ejemplo 1
# List comprehension para crear una lista con los números del 0 al 4.
numeros = [x for x in range(5)]
print(numeros)
# En esta los elementos que tenemos son:
# x: Es la expresión que toma el valor, en este caso de 1 al 5.
# for x in range(5): Es la expresión que hace que salgan los núemros del 0 al 4 mediante la utilización de un bucle for.

# Ejemplo 2
# List comprehension para crear una lista con los cuadrados de los números del 0 al 9.
cuadrados = [x**2 for x in range(10)]
print(cuadrados)
# En esta los elementos que tenemos son:
# x**2: Es la expresión que transforma cada x en su cuadrado.
# for x in range(10): Es la expresión que hace que se eleven al cuadrado los números del 0 al 9 mediante un bucle for.
# Ejemplo 2
````
