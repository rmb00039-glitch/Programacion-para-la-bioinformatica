### Ejercicio entrenamiento 1 ###
Dada la siguiente función que calcula la suma de cuadrados, escribe el código necesario para realizar pruebas de código usando doctest.

La suma de cuadrados se refiere a dado un numero n sumar todos los cuadrados hasta llegar a él, es decir, si n=3 la suma de cuadros sera $1^{2} + 2^{2} + 3^{2}$, $1+4+9$, =14`




````python
def sum_of_squares(n):
    """Calcula la suma de los cuadrados hasta n.

    >>> sum_of_squares(3)
    14
    >>> sum_of_squares(0)
    0
    >>> sum_of_squares(1)
    1
    >>> sum_of_squares(5)
    55
    >>> sum_of_squares(10)
    385
    >>> sum_of_squares(-1)
    Traceback (most recent call last):
    ...
    ValueError: n debe ser >= 0
    """

    if not n >= 0:
        raise ValueError("n debe ser >= 0")

    result = 0
    for i in range(n + 1):
        result += i ** 2

    return result

if __name__ == "__main__":
    import doctest
    doctest.run_docstring_examples(sum_of_squares, globals(), verbose=True)
````

### Ejercicio entrenamiento 2:

Dada la función complementaria_inversa (dna_secuencia) que devuelve la complementaria inversa de una secuencia de ADN dada, escribe tests unitarios utilizando unittest. Los test deben verificar si la función produce los resultados esperados para diferentes secuencias de ADN.

Recuerda tener en cuenta casos especiales y asegurarte de que la función maneje correctamente cualquier secuencia de ADN proporcionada.

La función debe seguir los pasos adecuados para encontrar el complemento inverso de la secuencia de ADN, teniendo en cuenta las bases complementarias (A-T, C-G) y el orden inverso de la secuencia.

Referencia:

    - Complementary DNA Strands:
    
    https://www.wikihow.life/Find-the-Reverse-Complement-of-a-DNA-Sequence#:~:text=Trace%20through%20the%20sequence%20backwards,last%20nucleotide%20in%20the%20sequence.&text=As%20you%20pass%20over%20each,hand%20side%20of%20the%20page.
    

````python
# Importar librerias requeridas
import unittest
import sys

def complementaria_inversa(dna_seq):
    """
    Encuentra la complementaria inversa de una secuencia DNA.

    Parametros:
    - dna_secuencia (str): Entrada DNA.

    Return:
    - str: Complementaria inversa del DNA.

    Referencia:
    - Complementary DNA Strands: https://www.wikihow.life/Find-the-Reverse-Complement-of-a-DNA-Sequence
    """
    complement = {"A": "T", "T": "A", "C": "G", "G": "C"}
    reverse_dna = dna_seq[::-1]
    complementaria_inversa = "".join(complement[base] for base in reverse_dna)
    return complementaria_inversa


# Tests
class TestEjercicio2(unittest.TestCase):
    def test_reverse_complement(self):
        # Casos normales
        self.assertEqual(complementaria_inversa("ATCG"), "CGAT")
        self.assertEqual(complementaria_inversa("AAAA"), "TTTT")
        self.assertEqual(complementaria_inversa("ACGT"), "ACGT")
        self.assertEqual(complementaria_inversa("GTCA"), "TGAC")

        # Caso con todas las bases
        self.assertEqual(complementaria_inversa("ATGCATGC"), "GCATGCAT")

        # Caso con una sola base
        self.assertEqual(complementaria_inversa("A"), "T")
        self.assertEqual(complementaria_inversa("T"), "A")
        self.assertEqual(complementaria_inversa("C"), "G")
        self.assertEqual(complementaria_inversa("G"), "C")

        # Cadena vacía
        self.assertEqual(complementaria_inversa(""), "")

    def test_invalid_input(self):
        # Caracteres inválidos en la secuencia
        with self.assertRaises(KeyError):
            complementaria_inversa("ATBX")

        # Números en la secuencia
        with self.assertRaises(KeyError):
            complementaria_inversa("AT1G")

        # Minúsculas (dependiendo de si la función las soporta)
        with self.assertRaises(KeyError):
            complementaria_inversa("atcg")


if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase(TestEjercicio2)
    unittest.TextTestRunner(verbosity=2, stream=sys.stderr).run(suite)
````

### Ejercicio entrenamiento 3
**Hablamos de tests en verde cuando todos nuestros tests se ejecutan correctamente y dan el resultado esperado y tests en rojo en caso contrario.**

En el siguiente ejercicio, escribe el código necesario que haga cumplir todos los tests, es decir, que los tests estén *en verde* **sin modificar en ningún momento la parte correspondiente al código de los tests** (`class TestMonster`).

````python
import unittest
import sys


class Monster:
    """A cute monster"""

    def __init__(self, name, size, basic_type='fire'):
        """Creates a new monster"""
        self.name = name
        self.size = size
        self.basic_type = basic_type.lower()

    def fight(self, monster2):
        """Two monsters fight each other. Returns the name of the monster winning
        or 'draw' otherwise.

        Monster battles have very specific rules:
        - Water wins over Fire
        - Electrical wins over Water
        - Plant wins over Electrical
        - Fire wins over Plant

        if both monsters fighting have the same type, bigger monster wins!
        if both monsters fighting have the same type and size, it's a draw

        Any other combination, it's a draw
        """
        # Definimos las reglas de victoria
        victory_rules = {
            'water': 'fire',
            'electrical': 'water',
            'plant': 'electrical',
            'fire': 'plant'
        }

        # Mismo tipo
        if self.basic_type == monster2.basic_type:
            if self.size > monster2.size:
                return self.name
            elif self.size < monster2.size:
                return monster2.name
            else:
                return 'draw'

        # Diferente tipo - verificamos las reglas
        if victory_rules.get(self.basic_type) == monster2.basic_type:
            return self.name
        elif victory_rules.get(monster2.basic_type) == self.basic_type:
            return monster2.name

        # Cualquier otro caso es empate
        return 'draw'


class TestMonster(unittest.TestCase):

    def test_create_a_new_monster(self):
        # Create a monster
        pika = Monster("Pika", 10, "electrical")

        # Test if it is not none
        self.assertIsNotNone(pika)

        # Test if attributes have been set properly
        self.assertEqual("Pika", pika.name)
        self.assertEqual(10, pika.size)
        self.assertEqual("electrical", pika.basic_type)

    def test_fight_water_fire(self):
        """Test which monsters wins in a water vs fire fight"""
        squirtle = Monster("Squirtle", 10, "water")
        char = Monster("Char", 10, "fire")

        self.assertEqual("Squirtle", squirtle.fight(char))

    def test_fight_fire_plant(self):
        """Test which monsters wins in a plant vs fire fight"""
        char = Monster("Char", 10, "fire")
        bulb = Monster("Bulb", 15, "plant")

        self.assertEqual("Char", char.fight(bulb))

    def test_electrical_electrical_same_size(self):
        """Test which monsters wins if both have same type and size"""
        pika = Monster("Pika", 10, "electrical")
        pichu = Monster("Pichu", 10, "electrical")

        self.assertEqual("draw", pika.fight(pichu))

    def test_electrical_electrical_different_size(self):
        """Test which monsters wins if both have same type but different size"""
        pika = Monster("Pika", 10, "electrical")
        pichu = Monster("Pichu", 9, "electrical")

        self.assertEqual("Pika", pika.fight(pichu))


if __name__ == '__main__':
    suite = unittest.TestLoader().loadTestsFromTestCase(TestMonster)
    unittest.TextTestRunner(verbosity=1, stream=sys.stderr).run(suite)


````

````python
import unittest
import doctest

def elevar_a_potencia(base, exponente):
    """
    Calcula la potencia de una base elevada a un exponente.

    Maneja el caso especial de exponente 0 (resultado 1).
    Lanza ValueError para 0 elevado a un exponente negativo.
    Permite que TypeError ocurra naturalmente si las entradas no son numéricas.

    >>> elevar_a_potencia(2, 3)
    8.0
    >>> elevar_a_potencia(5, 0)
    1.0
    >>> elevar_a_potencia(0, 5)
    0.0
    >>> elevar_a_potencia(2, -1)
    0.5
    >>> elevar_a_potencia(0, -2)
    Traceback (most recent call last):
        ...
    ValueError: Base cero con exponente negativo no está definido.
    >>> elevar_a_potencia(0, 0)
    1.0
    >>> elevar_a_potencia(2.5, 2)
    6.25
    >>> elevar_a_potencia("texto", 2)
    Traceback (most recent call last):
        ...
    TypeError: Las entradas deben ser numéricas o convertibles a números.
    """
    # Intentar convertir base y exponente a float.
    try:
        num_base = float(base)
        num_exponente = float(exponente)
    except ValueError:
        raise TypeError("Las entradas deben ser numéricas o convertibles a números.")

    # Caso especial: exponente es 0
    if num_exponente == 0:
        return 1.0

    # Caso especial: base es 0 y exponente es negativo
    if num_base == 0 and num_exponente < 0:
        raise ValueError("Base cero con exponente negativo no está definido.")

    # Cálculo general de la potencia
    return num_base ** num_exponente

class TestElevarAPotencia(unittest.TestCase):
    def test_enteros_positivos(self):
        self.assertEqual(elevar_a_potencia(2, 3), 8.0)

    def test_exponente_cero(self):
        self.assertEqual(elevar_a_potencia(5, 0), 1.0)

    def test_base_cero_exponente_positivo(self):
        self.assertEqual(elevar_a_potencia(0, 5), 0.0)

    def test_exponente_negativo(self):
        self.assertEqual(elevar_a_potencia(2, -1), 0.5)

    def test_error_base_cero_exponente_negativo(self):
        with self.assertRaisesRegex(ValueError, "Base cero con exponente negativo no está definido."):
            elevar_a_potencia(0, -2)

    def test_numeros_flotantes(self):
        self.assertEqual(elevar_a_potencia(2.5, 2), 6.25)

    def test_cero_elevado_a_cero(self):
        self.assertEqual(elevar_a_potencia(0, 0), 1.0)

    def test_type_error_entrada_no_numerica(self):
        with self.assertRaises(TypeError):
            elevar_a_potencia("texto", 2)

if __name__ == '__main__':
    doctest.run_docstring_examples(elevar_a_potencia, globals(), verbose=True)

    suite = unittest.TestLoader().loadTestsFromTestCase(TestElevarAPotencia)
    unittest.TextTestRunner(verbosity=1).run(suite)
````

````python
import unittest
import sys
import math

class Matrix2x2:
    def __init__(self, a=1., b=0., c=0., d=1.):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def __repr__(self):
        return f"Matrix2x2(a={self.a}, b={self.b}, c={self.c}, d={self.d})"

    def __str__(self):
        return f"[[{self.a}, {self.b}],\n [{self.c}, {self.d}]]"

    def __add__(self, other):
        if not isinstance(other, Matrix2x2):
            raise TypeError("Solo se pueden sumar matrices 2x2")
        return Matrix2x2(
            self.a + other.a,
            self.b + other.b,
            self.c + other.c,
            self.d + other.d
        )

    def __sub__(self, other):
        if not isinstance(other, Matrix2x2):
            raise TypeError("Solo se pueden restar matrices 2x2")
        return Matrix2x2(
            self.a - other.a,
            self.b - other.b,
            self.c - other.c,
            self.d - other.d
        )

    def __mul__(self, scalar):
        if not isinstance(scalar, (int, float)):
            raise TypeError("El escalar debe ser un número")
        return Matrix2x2(
            self.a * scalar,
            self.b * scalar,
            self.c * scalar,
            self.d * scalar
        )

    def determinant(self):
        return self.a * self.d - self.b * self.c

    def __eq__(self, other):
        if not isinstance(other, Matrix2x2):
            return False
        return (math.isclose(self.a, other.a) and
                math.isclose(self.b, other.b) and
                math.isclose(self.c, other.c) and
                math.isclose(self.d, other.d))

class TestMatrix2x2(unittest.TestCase):
    def test_creacion(self):
        # Matriz por defecto (identidad)
        m1 = Matrix2x2()
        self.assertEqual((m1.a, m1.b, m1.c, m1.d), (1., 0., 0., 1.))

        # Matriz con valores específicos
        m2 = Matrix2x2(1, 2, 3, 4)
        self.assertEqual((m2.a, m2.b, m2.c, m2.d), (1., 2., 3., 4.))

        # Matriz con valores extremos
        m3 = Matrix2x2(1e10, -1e10, 1e-10, -1e-10)
        self.assertEqual((m3.a, m3.b, m3.c, m3.d), (1e10, -1e10, 1e-10, -1e-10))

    def test_repr_str(self):
        m = Matrix2x2(1.5, 2.5, 3.5, 4.5)

        # Test __repr__
        self.assertEqual(repr(m), "Matrix2x2(a=1.5, b=2.5, c=3.5, d=4.5)")

        # Test __str__
        self.assertEqual(str(m), "[[1.5, 2.5],\n [3.5, 4.5]]")

    def test_suma(self):
        m1 = Matrix2x2(1, 2, 3, 4)
        m2 = Matrix2x2(5, 6, 7, 8)
        m_id = Matrix2x2()  # Identidad
        m_zero = Matrix2x2(0, 0, 0, 0)

        # Suma típica
        self.assertEqual(m1 + m2, Matrix2x2(6, 8, 10, 12))

        # Suma con identidad
        self.assertEqual(m1 + m_id, Matrix2x2(2, 2, 3, 5))

        # Suma con cero
        self.assertEqual(m1 + m_zero, m1)

        # Error: suma con no matriz
        with self.assertRaises(TypeError):
            m1 + "no_es_matriz"

    def test_resta(self):
        m1 = Matrix2x2(1, 2, 3, 4)
        m2 = Matrix2x2(5, 6, 7, 8)
        m_id = Matrix2x2()  # Identidad
        m_zero = Matrix2x2(0, 0, 0, 0)

        # Resta típica
        self.assertEqual(m1 - m2, Matrix2x2(-4, -4, -4, -4))

        # Resta con identidad
        self.assertEqual(m1 - m_id, Matrix2x2(0, 2, 3, 3))

        # Resta con cero
        self.assertEqual(m1 - m_zero, m1)

        # Error: resta con no matriz
        with self.assertRaises(TypeError):
            m1 - "no_es_matriz"

    def test_multiplicacion_escalar(self):
        m = Matrix2x2(1, 2, 3, 4)

        # Multiplicación por positivo
        self.assertEqual(m * 2, Matrix2x2(2, 4, 6, 8))

        # Multiplicación por negativo
        self.assertEqual(m * -1, Matrix2x2(-1, -2, -3, -4))

        # Multiplicación por cero
        self.assertEqual(m * 0, Matrix2x2(0, 0, 0, 0))

        # Multiplicación por 1
        self.assertEqual(m * 1, m)

        # Error: multiplicación por no escalar
        with self.assertRaises(TypeError):
            m * "no_es_escalar"

    def test_determinante(self):
        # Matriz típica
        m1 = Matrix2x2(1, 2, 3, 4)
        self.assertEqual(m1.determinant(), -2)

        # Matriz singular
        m2 = Matrix2x2(1, 2, 2, 4)
        self.assertEqual(m2.determinant(), 0)

        # Matriz identidad
        m_id = Matrix2x2()
        self.assertEqual(m_id.determinant(), 1)

        # Valores decimales
        m3 = Matrix2x2(1.5, 2.5, 3.5, 4.5)
        self.assertAlmostEqual(m3.determinant(), 1.5*4.5 - 2.5*3.5)

    def test_igualdad(self):
        m1 = Matrix2x2(1, 2, 3, 4)
        m2 = Matrix2x2(1, 2, 3, 4)
        m3 = Matrix2x2(1.0000001, 2.0000001, 3.0000001, 4.0000001)
        m4 = Matrix2x2(5, 6, 7, 8)

        # Igualdad exacta
        self.assertEqual(m1, m2)

        # Igualdad con valores muy cercanos (usando math.isclose)
        self.assertEqual(m1, m3)

        # Desigualdad
        self.assertNotEqual(m1, m4)

        # Comparación con no matriz
        self.assertNotEqual(m1, "no_es_matriz")

if __name__ == '__main__':
    unittest.main()
````

````python
import unittest
import sys
class Peptide:
    # Diccionario de aminoácidos estándar con sus códigos de una letra
    STANDARD_AMINO_ACIDS = {
        'A': 'Alanine',
        'R': 'Arginine',
        'N': 'Asparagine',
        'D': 'Aspartic Acid',
        'C': 'Cysteine',
        'Q': 'Glutamine',
        'E': 'Glutamic Acid',
        'G': 'Glycine',
        'H': 'Histidine',
        'I': 'Isoleucine',
        'L': 'Leucine',
        'K': 'Lysine',
        'M': 'Methionine',
        'F': 'Phenylalanine',
        'P': 'Proline',
        'S': 'Serine',
        'T': 'Threonine',
        'W': 'Tryptophan',
        'Y': 'Tyrosine',
        'V': 'Valine'
    }

    def __init__(self, sequence):
        """
        Inicializa un péptido con la secuencia dada.
        Lanza ValueError si la secuencia contiene aminoácidos no estándar.
        """
        self.sequence = sequence.upper()  # Convertir a mayúsculas para uniformidad
        self._validate_sequence()

    def _validate_sequence(self):
        """Valida que la secuencia solo contenga aminoácidos estándar."""
        for aa in self.sequence:
            if aa not in self.STANDARD_AMINO_ACIDS:
                raise ValueError(f"El aminoácido '{aa}' no es estándar")

    def __repr__(self):
        """Representación oficial del péptido."""
        return f"Peptide('{self.sequence}')"

    def __str__(self):
        """Representación legible del péptido."""
        return self.sequence

    def __add__(self, other):
        """
        Concatena dos péptidos.
        Lanza TypeError si el otro objeto no es un Peptide.
        """
        if not isinstance(other, Peptide):
            raise TypeError("Solo se pueden concatenar péptidos con péptidos")
        return Peptide(self.sequence + other.sequence)

    def polyh(self, count):
        """
        Añade un número específico de histidinas (H) al final del péptido.
        Lanza ValueError si count no es un entero positivo.
        """
        if not isinstance(count, int) or count < 0:
            raise ValueError("El número de histidinas debe ser un entero positivo")
        return Peptide(self.sequence + 'H' * count)

# Ejemplos de uso:
if __name__ == '__main__':
    try:
        # Creación de péptidos
        pep1 = Peptide("ARARAR")
        pep2 = Peptide("TGGG")

        # Concatenación
        pep3 = pep1 + pep2
        print(f"Concatenación: {pep3}")  # Debería imprimir: ARARARTGGG

        # Añadir histidinas
        pep4 = pep3.polyh(5)
        print(f"Con histidinas añadidas: {pep4}")  # Debería imprimir: ARARARTGGGHHHHH

        # Prueba con aminoácido no estándar (debería lanzar ValueError)
        bad_pep = Peptide("PPXDNR")
    except ValueError as e:
        print(f"Error: {e}")
````
