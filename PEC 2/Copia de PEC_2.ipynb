###Ejercicio entrenamiento 5

Escribe un programa que dados dos diccionarios con posibles claves repetidas y valores por cada clave de tipo numérico, devuelva un nuevo diccionario con el contenido de los dos diccionarios originales y, si existe una clave repetida, el valor de esa clave sea la media de los valores originales:

````python
diccionario_1 = {'melón': 1.5, 'sandía': 3.5, 'uva': 1.0, 'mandarina': 3.1}
diccionario_2 = {'melón': 2.5, 'sandía': 1.5, 'platano': 3.0, 'naranja': 4.3}

def media_diccionarios(d1, d2):
   # A completar
   resultado = {}
   todas_las_claves = set(d1.keys()) | set(d2.keys())
   for clave in todas_las_claves:
    if clave in d1 and clave in d2:
      resultado[clave] = (d1[clave] + d2[clave]) / 2
    elif clave in d1:
      resultado[clave] = d1[clave]
    else:
      resultado[clave] = d2[clave]
   return resultado

print(media_diccionarios(diccionario_1, diccionario_2))
````

````python
# Masas atómicas
masas = {'H': 1.007825, 'C': 12.01, 'O': 15.9994,
         'N': 14.0067, 'S': 31.972071, 'P': 30.973762}


def calcula_masa_atomica(molecula):
    """Calcula la masa atomica de una molecula"""
    # Necesitamos una variable donde acumular todas las masas parciales
    # que vayamos calculando en el siguiente bucle:
    masa = 0.0
    grupos = molecula.split('-')

    for grupo in grupos:
        atomo = grupo[0] # Extraemos la letra correspondiente al átomo
        cantidad = grupo[1:] # Extraemos el número de átomos
# El siguiente bucle if es para que si no hay catidad, decir a python que lo qu ehay es un átomo, no cero
        if not cantidad:
            cantidad = 1
        else:
            cantidad = int(cantidad)
        masa += masas[atomo] * cantidad
    return round(masa, 4)
````

### Ejercicio 5 **<font color="blue" size="+2">EG</font>**

Uno de los algoritmos más básicos en criptografía es el cifrado César (https://es.wikipedia.org/wiki/Cifrado_C%C3%A9sar), el cual fue utilizado por Julio César para comunicarse con sus generales. Este algoritmo consiste en, dado un texto, por cada letra del texto, aplicar un desplazamiento para obtener una nueva letra diferente de la original.

Ejemplo:
Si asignamos el número 1 a la primera letra del abecedario, "A", 2 a la siguiente, "B", y así sucesivamente, considera el siguiente mensaje:
<code>
ABC 123
</code>

Si aplicamos un desplazamiento de 3, el resultado sería:
<code>
DEF 456
</code>

Esto sucede porque, al sumar un desplazamiento de 3 a "ABC", obtenemos "DEF". También se pueden aplicar desplazamientos negativos. Por ejemplo, con un desplazamiento de -1 en el mensaje "ABC", el mensaje cifrado resultante sería "ZAB".

Desarrolla una función que, **dado un mensaje original y un desplazamiento, calcule y devuelva el mensaje cifrado**:

````python
"""
Podemos simplificar sin problemas el uso de palabras solo
a mayúsculas y al abecedario en inglés.
Vamos a utilizar la propiedad de que la primera letra del
abecedario A se corresponde con el entero
65 en ASCII y Z con el 90, por lo tanto, tenemos una longitud
de abecedario de 90-65+1=26 letras.
"""

def cifrado_cesar(mensaje, desplazamiento=1):

    """
    Cifra el mensaje utilizando el metodo de
    Cesar dado un desplazamiento
    """
    mensaje_cifrado = ""

    # Codigo a completar
    for caracter in mensaje:
      if 'A' <= caracter <= 'Z':
        nueva_letra = chr((ord(caracter) - 65 + desplazamiento) % 26 + 65)
        mensaje_cifrado += nueva_letra
      else:
        mensaje_cifrado += caracter

    return mensaje_cifrado
````

````python
%%ipytest
# Test Ejercicio 5
def test_cifrado_cesar():
  assert cifrado_cesar('ABCD', 27) == "BCDE"
  assert cifrado_cesar('ABCD', 1) == "BCDE"
  assert cifrado_cesar('XYZ', 1) == "YZA"
  assert cifrado_cesar("PROGRAMACION PARA LA BIOINFORMATICA", 26) == "PROGRAMACION PARA LA BIOINFORMATICA"
  assert cifrado_cesar("PROGRAMACION PARA LA BIOINFORMATICA", 6) == "VXUMXGSGIOUT VGXG RG HOUOTLUXSGZOIG"
````

````python
def first_last_same(numberList):
    print("Dada la lista:", numberList)

    # Codigo a completar
    if len(numberList) == 0:
        return False  # Da fale si la lista esta vacia
    return numberList[0] == numberList[-1]  # Compara el primer y último número de la lista
print(first_last_same([10, 20, 30, 10]))
print(first_last_same([10, 20, 30, 40, 11]))
````

### Ejercicio 9 **<font color="green" size="+2">MU</font>**

A partir de dos listas de números, genera una nueva lista que contenga únicamente los números **impares** presentes en ambas listas.

````python
def merge_list(list1, list2):
    result_list = []

    # Codigo a completar
    # Obtenemos ambas lostas conjuntas
    listas_conjuntas = list1 + list2
    # Filtramos solo los números impares
    for num in listas_conjuntas:
        if num % 2 != 0:  # Mete al número en la lista si este es impar
            result_list.append(num)
    return result_list
````

````python
import math
def distance(x1, y1, z1, x2, y2, z2):
    """Calcula la distancia euclidiana entre dos puntos 3D"""
    d = 0

    # Codigo a completar
    d = math.sqrt((x2 - x1)**2 + (y2 - y1)**2 + (z2 - z1)**2) #Utilizamosla misma fórmula y un código muy similar al del ejericcio de ejemplo 1

    return d
````
